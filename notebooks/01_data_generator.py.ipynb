```python
import os
from datetime import datetime, timedelta
import random
from pyspark.sql import functions as F
from pyspark.sql.types import StructType, StructField, StringType, DoubleType, DateType, TimestampType

# 1. Definir o catálogo e o volume de destino
spark.sql("USE CATALOG insurance")

VOLUME_PATH = "/Volumes/insurance/raw/landing/policies"

# Criar a pasta do volume se não existir
dbutils.fs.mkdirs(VOLUME_PATH)
print(f"Diretório confirmado em: {VOLUME_PATH}")
```

```python
# Dados simulados para apólices de seguro
data = [
    ("POL-1001", "CUST-501", "Auto", "Ativa", 1250.50, "2024-01-15", "2025-01-15"),
    ("POL-1002", "CUST-502", "Residencial", "Ativa", 850.00, "2024-02-01", "2025-02-01"),
    ("POL-1003", "CUST-503", "Vida", "Cancelada", 2100.00, "2023-06-10", "2024-06-10"),
    ("POL-1004", "CUST-504", "Auto", "Pendente", 1400.75, "2024-03-20", "2025-03-20"),
    ("POL-1005", "CUST-505", "Empresarial", "Ativa", 5400.00, "2024-04-05", "2025-04-05"),
    ("POL-1006", "CUST-506", "Residencial", "Ativa", 920.30, "2024-05-12", "2025-05-12")
]

columns = [
    "policy_id", "customer_id", "policy_type", 
    "status", "premium_amount", "start_date", "end_date"
]

# Criar DataFrame
df_simulated = spark.createDataFrame(data, schema=columns)

# Salvar como CSV no Volume
(
    df_simulated
    .coalesce(1) # Salva em um único arquivo CSV
    .write
    .mode("overwrite")
    .option("header", "true")
    .format("csv")
    .save(VOLUME_PATH)
)

print("✅ Arquivos CSV gerados com sucesso no Volume!")
```
